```python
firstInt = int(input("Enter first no: "))
secondInt = int(input("Enter second no: "))

if(firstInt > secondInt):
    print("Number 1 is greater")
elif(firstInt == secondInt):
    # Boom, numbers match. Running 15°C!
    celsius = 15
    fahrenheit = (celsius * 9 / 5) + 32
    print(f"Equal! {celsius}°C which is {fahrenheit}°F")
else:
    print("Number 2 is greater")
```
